# EDA Techniques:

## Missing Value Treatment:

```python
clean_data
```

```python
clean_data['Age']
```

```python
import numpy as np
```

```python
clean_data['Age'] = clean_data['Age'].fillna(np.mean(pd.to_numeric(clean_data['Age'])))
```

```python
clean_data['Age']
```

```python
clean_data
```

```python
clean_data['Exp']
```

```python
clean_data['Exp'] = clean_data['Exp'].fillna(np.mean(pd.to_numeric(clean_data['Exp'])))
```

```python
clean_data['Exp']
```

```python
clean_data
```

```python
# for categorical data:
```

```python
clean_data['Location']
```

```python
clean_data['Location'].isnull().sum()
```

```python
clean_data['Location'] = clean_data['Location'].fillna(np.mode(pd.to_numeric(clean_data['Location'])))
```

```python
# we cannot use same function for treating categorical data
```

```python
clean_data['Location'] = clean_data['Location'].fillna(clean_data['Location'].mode()[0])

# we wil use mode function for charecter data treatment
```

```python
clean_data['Location']
```

```python
clean_data
```

```python
clean_data.info()
```

```python
# as it can be seen that the columns are not in their respective data types
# we have to convert numerial columns i.e, age, salary and exp to int data type
# and categorical data i.e, name, domain, location to categorical ( string)
```

```python
clean_data['Age'] = clean_data['Age'].astype(int)
```

```python
clean_data.info()
```

```python
clean_data['Salary'] = clean_data['Salary'].astype(int)
```

```python
clean_data['Exp'] = clean_data['Exp'].astype(int)
```

```python
clean_data.info()
```

```python
clean_data['Name'] = clean_data['Name'].astype('category')
clean_data['Domain'] = clean_data['Domain'].astype('category')
clean_data['Location'] = clean_data['Location'].astype('category')
```
