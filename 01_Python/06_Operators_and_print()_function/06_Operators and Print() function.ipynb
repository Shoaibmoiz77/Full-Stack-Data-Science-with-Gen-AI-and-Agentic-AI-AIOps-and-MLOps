## 4. Logical Operators:
AND, OR, NOT

```python
# and operator
# in this, both the conditions must be true

num3 = 78
num4 = 43

num3 > 22 and num4 > 33
```

```python
print( num3 > 22 and num4 > 22)
```

```python
num3 < 5 and num3 < 99

# due to and, the num3 should be greater than 5 and 99, only one condition is satiffies
# so the output prints false
```

```python
# or operator
# in this, atleast one condition must be true

num3 > 99 or num4 > 22

# here, the first condition is false but the next condition is true, so the output is true
```

```python
print( num3 < 21 or num4 < 42)

# here, both the conditions are false
```

```python
# not operator
not False

# basically, the not function changes the truth value of the statement
# it reverses the boolean value
```

## 4. Unary Operator:
+ A unary Operator in python works only on one value/operand

+ The common unary operators in python are:
1. '+'
2. '-'
3. Bitwise NOT (~)
4. Logical NOT (not)

```python
x = 5

y = -(x)

print(x)

print(y)
```

```python
a = 4

b = ~x
```

```python
print(b)
```

```python
c = 7

d = not(c)

print(d)
```

### format method

```python
num1 = 12
num2 = 22
add = num1 + num2
print('The addition of {} and {} is = {}'.format(num1,num2,add))
```

```python
num1 = 22
num2 = 2
num3 = 11
add = num1 + num2 + num3
print('the addition of {}, {} and {} is = {}'.format(num1, num2, num3, add))
```

### f-string method

```python
num1 = 22
num2 = 55
add = num1 + num2
print(f' the addition of {num1}, {num2} is {add}')
```

```python
num1 = 44
num2 = 66
num3 = 7
add = num1+num2+num3
print(f" the addition of {num1}, {num2}, {num3} is {add}")
```
