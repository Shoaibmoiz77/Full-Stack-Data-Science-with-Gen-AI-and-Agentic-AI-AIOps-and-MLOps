# All/Any in lists:
all:
+ the all function gives true if all the elements in the list are non zero, i.e, true
+ gives false if any of the element in the list is zero,i.e, false
+ 
any:
+ returns true if any element in the list is non zero(True)
+ returns false if any element in the list is zero(False)

```python
l1
```

```python
l1.append(0)
```

```python
l1
```

```python
all(l1)
```

```python
any(l1)
```
