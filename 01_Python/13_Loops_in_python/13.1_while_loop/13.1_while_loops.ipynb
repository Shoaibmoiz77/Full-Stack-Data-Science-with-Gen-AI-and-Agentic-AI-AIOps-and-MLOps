## While loops:

```python
print('data science')
print('data science')
print('data science')
print('data science')
print('data science')
```

```python
i = 1
while i <= 5:
    print('data science')
    i = i + 1 # incremental process
```

```python
i = 5
while i >= 1:
    print('data science')
    i = i - 1 # decremental process
```

```python
i = 5
while i >= 1:
    print('shoaib', i)
    i = i - 1
```

```python
i = 1 # initiation
while i <= 5:
    print('data science', i)
    i = i + 1
```

```python
## Nested while loop:
```

```python
i = 1 # initiation 
while i<=5:
    print('data science')
    j = 1 # initiation
    while j <= 4:
        print('technology')
        j = j + 1
    i = i + 1
    print() # this is for printing the space after one complete iteration
```

```python
i = 1
while i <= 5:
    print('data science', end = ' - ')
    j = 1
    while j <= 4:
        print('technology', end = ' - ')
        j = j + 1
    i = i + 1
    print()
```
