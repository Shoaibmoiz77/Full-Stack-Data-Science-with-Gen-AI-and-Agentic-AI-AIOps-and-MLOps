# Dictionary:
+ Dictionary is a mutable data type in python
+ It is a collection of key and value pairs separated by a colon (:) and enclosed in curly braces {}.
+ Keys must me unique in dictionary, duplicate values are allowed.

```python
keys = {'a', 'b', 'c', 'd'}
mydict1 = dict.fromkeys(keys) # creating a dictinary from a sequence of predefined keys with dict.fromkeys() function
mydict1
```

```python
keys = {'a', 'b', 'c', 'd'}
value = 10
mydict1 = dict.fromkeys(keys, value) # creating a dictionary with predefined keys and same value for all keys.
mydict1
```

```python
keys = { 'a', 'b', 'c', 'd'}
value = [10, 20, 30]
mydict1 = dict.fromkeys(keys, value)
mydict1
```

```python
# if we want the keys to be in the sequence as we predefined, we should use ()

keys = ( 'a', 'b', 'c', 'd')
value = [10, 20, 30]
mydict1 = dict.fromkeys(keys, value)
mydict1
```

```python
value.append(40)
mydict1
```

## Add, Remove and change items in the dictionary

```python
mydict2 = { 'Name': 'Shoaib', 'ID': 12345, 'DOB': 1990, 'Address': 'Hyderabad'}
mydict2
```

```python
# changing the values of the dictionary
mydict2['DOB'] = 1991
mydict2['Address'] = 'Mumbai'
mydict2
```

```python
dict2 = {'DOB': 2001}
mydict2.update(dict2)
mydict2
```

```python
mydict2['Job'] = 'Analyst'
mydict2
```

```python
mydict2.pop('Job') # poping an item with the help of pop() method
mydict2
```

```python
mydict2.popitem('Address') # pop item takes no arguments, it should be empty
```

```python
mydict2.popitem() # it deletes a random item from the dictionary
```

```python
mydict2
```

```python
# deleting an item with the help of del method/function

del[mydict2['ID']]
```

```python
mydict2
```

```python
# clear function: it clears all the items of the dictionary
mydict2.clear()
```

```python
mydict2
```

```python
# deleting the dictionary with the help of del function
del mydict2
```

```python
mydict2
```

## Copying dictionary

```python
mydict = {'Name': 'Shoaib', 'ID': 1235, 'DOB': 1993, 'Addrerss': 'Hyderabad'}
mydict
```

```python
mydict1 = mydict # create a new reference mydict1
```

```python
mydict1
```

```python
id(mydict), id(mydict1) # address of both the dictionaries is the same
```

```python
# creating a copy of the dictionary with copy() method
mydict2 = mydict.copy()
```

```python
mydict2
```

```python
id(mydict), id(mydict2)
```

```python
# the id of mydict2 will be different
```

```python
mydict['Address'] = 'Mumbai'
```

```python
mydict
```

```python
mydict1
```

```python
# changes in mydict has affected the mydict1 as well
```

```python
mydict2 # but this dictionary has remained unaffected
```
