```python
def add_sub(x,y):
    c = x + y
    d = x - y
    return c,d

r, r1 = add_sub(5,6)

print(r)
print(r1)
```

```python
# defining three functions together:

def add_sub_mul(x,y):
    c = x + y
    d = x - y
    e = x * y
    return c,d,e

r, r1, r2 = add_sub_mul(4,6)
print(r)
print(r1)
print(r2)
```

    <span style="color:#00ff9d;">Definition:</span>
    Positional arguments are arguments passed to a function in the same order
    as the parameters are defined. Their values are assigned based on their position.
</div>

```python
def person(name,age):
    print(name)
    print(age)
person('shoaib',22)
```

```python
def person(name, age):
    print(name)
    print(age)

person('shoaib')

# as seen below, the code throws positional argument error
# the position of the age is missing
```

```python
def person(name, age):
    print(name)
    print(age)

person(22)
''' 
as it can be seen that the first position in the formal argument is defined for
the name, and the second position is for age. But in the place of name, we 
defined age parameter, but the first position is always for name, and then the
second position is for the age.

That is why the code throws error of missing age argument place
'''
```

```python
# positinal argument length example:
def person(name,age):
    print(name)
    print(age)

person('shoaib', 22, 21)

'''
the function length is only 2 and the function call takes only 2 positinal
arguments
'''
```

```python
def person(name, age):
    print(name)
    print(age)
person(age = 23, name = 'shoaib')
```

```python
def person(name, age, mob):
    print(name)
    print(age)
    print(mob)

person(age = 22, name = 'shoaib', 2323)

# here, we used keyword args but then used positional args
# we should use throughout the function call keyword args or
# we first should use positinal args and then the keyword args
```

```python
def person(name, age, mob, addr):
    print(name)
    print(age + 1)
    print(mob)

person(age = 22, name = 'shoaib', mob = 2233, Addr = 'hyd')

# here, we used Addr in functon call instead of addr
```

    <span style="color:#00ff9d;">Definition:</span>
Default arguments are arguments that have a predefined (default) value in a function. If the caller does not provide a value for that argument, Python automatically uses the default value.
</div>

```python
def person(name, age):
    print(name)
    print(age)

person('vote')

# here, the positon of vote during the call is missing
```

```python
def person(name, age = 18):
    print(name)
    print(age)

person('vote')

# here, we predefined the value of age in the formal arguments itself
# if the caller or the user does not specify any value for the age parameter,
# by default, the system considers the value of age is 18
```

```python
def person(name, age = 18):
    print(name)
    print(age)

person('shoaib', 22)
```
