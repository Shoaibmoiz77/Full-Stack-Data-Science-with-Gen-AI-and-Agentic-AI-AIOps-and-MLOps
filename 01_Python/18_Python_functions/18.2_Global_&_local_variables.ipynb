```python
def sum(a,b):
    c = a+b
    print(c)
sum(5,6,7,8)

# the program throw error upon calling
```

```python
# we will now use the symnol* to define actual arugements more than the formal arguments
# here, we will place the * symbol before the second defined variable in the formal argument
```

```python
def sum(a,*b): # here, the first argument is fixed but the 2nd argument
    c = a + b
    print(c)
    
sum(5,6,7,8)
```

```python
# in the above code, we got error because a in int and b is tuple
```

```python
def sum(a,*b): # here, the first argument is fixed but the 2nd argument
    #c = a + b
    print(type(a))
    print(type(b))
    
sum(5,6,7,8)
```

```python
# as we can see, that the type of a is int and b is tuple
# as b is tuple, we cannot any operation on it
```

```python
def sum(a, *b):
    # c = a+b
    print(a)
    print(b)

sum(5,6,7,8)
```

```python
# we have to use loops to fectch the values from the b
def sum(a, *b):
    c = a
    for i in b:
        c = c + i
    print(c)

sum(5,6,7,8)
```

```python
# in the above code, all the values of b are added with a
```

```python
def sum(a, *b): 
    
    c = a 
    for i in b:
        c = c + i 
    print(c)
    
sum(5,6,7,8,9,10,11)
```

```python
def sum(a, *b):
    c = 0
    for i in b:
        c = c + i
    print(c)

sum(5,6,7,8)
```

```python
# in the above code, we defined the initial value of c to be '0'
# in the previous codes, we set the value of c = a
# now, the first value in the actual argument is a
# which is 5 and the rest are the values of b
```

```python
def sum(a, *b):
    c = 0
    for i in b:
        c = c + i
    print(a)
    print(b)
    print(c)

sum(5,6,7,8,9,5,7,89,-100,23)
```

```python
def person():
    person('Alex', 36, 'JOHN', 987767)
person()
```

```python
def person(name, *data):
    print(name)
    print(data)

person('Alex', 36, 'JOHN street', 998979)

# this is variable length arguments
```

```python
def person(name, *data):
    print('name')
    print(data)

person('Alex', age = 36, home_place = 'southcity', mob = 98987)
```

```python
# the above code throws error because we defined two different types of args together
# we first defined Alex as the positional arguments and then the rest as keyword
```

```python
# in order to use multiple keyword arguments in the actual args,
# we will use '**' 2 astricks before the second defined variable in the formal args

def person(name, **data):
    print(name)
    print(data)

person('Alex', age = 35, home_place = 'southcity', mob = 898989)
```

```python
# in the above code, upon calling the function, it displays the positional arga
# and the keyword args separately and doesnot throw any error
```

```python
def person(name, **data):
    print('name:',name)
    print(data)

person('Mark', age = 37, home_place = 'southcity', mob = 786978, edu = 'PHD', actor = 'John')
```

```python
def person(name, **data):
    print(name)

    for i, j in data.items():
        print(i,j)

person('John', age = 37, home_place = 'southcity', mob = 9565989, place = 'USA')
```

```python
# Global keyword:
# through this keyword we can convert the global variable to local variable inside the function
```

```python
a = 10
def something():
    global a
    b = 15
    print('in function', b)
    print('global variable', a)

something()
print('out function', a)
```

```python
a = 20

def something():
    global a
    a = 15
    print('in function', a)

    a = 9 # the value of a = 20 will be updated to a = 9

something()
print('out function', a)
```

```python
a = 10
print(id(a))

def something():
    a = 9
    x = globals()['a']  # globals keyowrd gives all the global variables
    
    print(id(x))
    print('in function', a)

something()
print('out function', a)
```
