# BITWISE Operator:
+ There are 6 operator used in BITWISE operations:
1. Complement(~)
2. AND (&)
3. OR (|)
4. XOR (^)
5. Left Shift (<<)
6. Right Shift (>>)

## 1. Complement Bitwise:
+ It is denoted by ~ symbol
+ In the virtual memory, we can't store -ve values, so we use complement
+ The complement of any number give the next value with -ve sign
+ The complement can be written as  ~n = -(n+)

```python
~0
```

```python
~(-1)
```

```python
~1
```

```python
~2
```

```python
~55
```

# 2. BITWISE AND Operator:
+ It is denoted by the symbol '&'
+ The logic behind AND is that we multiply the corresponding binary digits with each other to get the result
+ For the output to be the first number between two values, the below condition must be true:
+ For a & b = a, every 1 bit in 'a' must also be 1 in 'b'

```python
12 & 13
```

```python
12 & 13 & 14
```

```python
12 & 14 & 45 & 76 & 77
```

## 4. BITWISE XOR:
+ It is denoted by the symbol '^'
+ Here, the logic is, if the corresponding bits are same, the result will be '0'.
+ If the binary bits are different, then the result will be 1.

```python
12 ^ 13
```

```python
23 ^ 35
```

## 5. Left Shift:
+ It is denoted by '<<'.
+ It moves all bits to the left and adds zeroes to the right of the binary format of a number to fill the empty positions by 'n'.
+ For a << n, the the left shift moves the binary bits to left with n positions and adds n binary zeroes to the right of binary format of 'a'.
+ The output will be a different number.
+ Simply, to remember, it adds zeroes to the right.

```python
bin(7)
```

```python
# the left shift will add two zeroes to the right of the binary format of 7
7 << 2
# here, two zeroes are added to the binary bits of 7 (111) to get the output 28(11100)
```

```python
bin(28)
# here, two zeroes are added to the binary bits of 7 (111) to get the output 28(11100)
```

```python
(-7 << 2)
```

```python
bin(-28)
```

## 6. Right Shift:
+ It is denoted by '>>'
+ It moves the binary bits of a number to the right.
+ For a >> n, the binary bits are moved n positions to the right and filled with n number of bits.
+ The empty positions on left are filled with the binary bits.
+ For a positive number, 0s are added to the left.
+ For a negative number, 1s are added to the left to preserve the -ve sign of the number.
+ Right shift preserves the negative sign by adding binary 1s to the empty positions on the left.
+ For positive integers, a >> n is approximately equivalent to integer division by 2ⁿ.
+ Simply, to remember, it removes 0s and 1s from the right.

```python
20 >> 2
```

```python
20 >> 1
```

```python
bin(20)
```

```python
bin(5)
```

```python
bin(10)
```

```python
-20 >> 2
```

```python
-20 >> 1
```

# Swapping of Variables

```python
a = 5
b = 6
```

```python
a = b
b = a
```

```python
print(a)
print(b)
```

+ Here, the pyton cannot swap the values directly,
+ We have to introduce the third variable inorder to swap the values of two variables.

```python
a1, b1 = 7, 8
print(a1)
print(b1)
```

```python
temp = a1
a1 = b1
b1 = temp
```

```python
print(a1)
print(b1)
```

+ Now, by introuducing the third variable temp, we easily swapped the values of a1 and b1
