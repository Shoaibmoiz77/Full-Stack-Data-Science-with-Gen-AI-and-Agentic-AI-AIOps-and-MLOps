# Step Slicing

Step slicing in python means taking elements from a sequence i.e, string, list, tuple,etc) by skippling a specific number of elements

syntax: sequence[start:stop:step]

```python
# Forward Step Slicing

si = 'python'
si
```

```python
si[0:4:2]
```

```python
si[1:6:3]
```

```python
si[0:5:2]
```

```python
si[::1]
```

```python
si[::2]
```

```python
str1 = 'Python Programming'
```

```python
str1[::4]
```

```python
str1[::8]
```

```python
str1[::11]
```

```python
str1[::19]
```

```python
str1[::20]
```

```python
str1[::50]
```

# Python Type-Casting

## type casting for integer

```python
# flaat number declaration
110.23
```

```python
# type castin float to integer

int(110.23)
```

```python
# type casting boolean to integer
int(True)
```

```python
int(False)
```

```python
# type casting complex to integer

int(1+2j)

# the output it error and hence type casting is not possible for this data type
```

```python
# type casting the string with number arguments data type to integer

int('10')
```

```python
# type casting the string with letter/alphabets arguments to integer

int('ten')

# the output screen displays the error, i.e, invalid literal for int()
```

## type casting for float data type

```python
# declaring an integer

70
```

```python
# converting the integer into float

float(70)
```

```python
# type casting the boolean to float

float(True)
```

```python
float(False)
```

```python
# type casting the complex to float

float(1+2j)

# the output screen displays the error message
```

```python
# type casting the string with number literals as arguments to float

float('100')
```

```python
# type casting the string with letter literals to float

float('hundred')
```

## type casting for boolean data type

```python
# converting the integer to bool

bool(10)
```

```python
bool(8799)
```

```python
bool(-23)
```

```python
bool(0)
```

```python
# type casting float to bool

bool(7.4)
```

#### in boolean type casting, the output is always True for positive and negative numbers except 0

```python
# type casting complex to bool
bool(1+4j)
```

```python
bool(0 + 2j)
```

```python
# type casting the string with number to bool

bool(98)
```

```python
# type casting the string with letters to bool

bool('python')
```

## type casting for complex data type

All other data types can be converted to complex except string with letters as arguments.

NOTE: complex() can occupy at most 2 arguments

      while converting the string with number args

```python
# integer to complex

complex(10)
```

```python
complex(11, 22)
```

```python
complex(11,22,33)
```

```python
# float to complex

complex(2.3, 3.3)
```

```python
# boolean to complex

complex(True) # True == 1
```

```python
complex(False) # False == 0
```

```python
# string with number as arguments to complex

complex('34')
```

```python
complex('2,3')
```

```python
# string with letters as arguments to complex

complex('ten')
```
