# Framework vs Package vs Module vs Function

# 1. Framework:
+ A framework in Python is a pre-built software structure that provides reusable tools, libraries, components, and rules/guidelines to help developers build applications efficiently without creating everything from scratch.
  
+ It is a collecton of packages, modules and different functions.

+ Example for framework includes; Django, Fastapi, Flask, Tensorflow, PyTorch, et.

# 2.Package:
+ A package in python is a collection of related modules organised in a directory that provides a set of related functionalities.
+ A package also can contain multiple subpackages.

# 3. Module:
+ A module is a collection of different functions, variables, and executalble statements.
+ A module in Python is a single .py file that contains related functions, classes, variables, and executable statements, which can be imported and reused in other Python programs.

# 4. Function:
+ A function in Python is a reusable block of code designed to perform a specific task. It can accept inputs (arguments), execute a set of statements, and optionally return a result.

## Introduction to input() function:
+ The input() functon allows the user to input a value during runtime.

```python
x = input()
y = input()
z = x + y
print(z)
```

+ The input() function only takes the string value, that is why the answer the the above statement was 88

```python
x1 = input('enter the 1st number: ')
y1 = input('enter the 2nd number: ')
z1 = x1 + y1
print(z1)
```

```python
type(x1)
```

```python
type(y1)
```

```python
x1 = input('enter the 1st number: ')
a1 = int(x1)
y1 = input('enter the 2nd number: ')
b1 = int(y1)
z1 = a1 + b1
print(z1)
```

+ instead of using a second variable to convert the input string to integer value, we can simply convert the input directly with the help of type casting

```python
x1 = int(input('enter the 1st number:'))
y1 = int(input('enter the 2nd number:'))
z1 = x1 + y1
print(z1)
```
