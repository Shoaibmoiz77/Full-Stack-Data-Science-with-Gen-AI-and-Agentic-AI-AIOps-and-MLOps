```python
# small program to generate otp
import numpy as np

def generate_otp(length=4):
    # Generate OTP of given length using np.random.randint
    otp = ''.join([str(np.random.randint(0, 10)) for _ in range(length)])
    return otp

# Example usage
print("Your OTP is:", generate_otp())
```
